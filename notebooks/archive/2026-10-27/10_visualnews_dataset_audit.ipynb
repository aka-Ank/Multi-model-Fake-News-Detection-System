# VisualNews Dataset Audit

This notebook audits the local article archives without bulk extraction or image downloading. It keeps VisualNews provenance separate from Fakeddit misinformation labels.

```python
import json
from pathlib import Path

import pandas as pd

report_dir = Path('../reports/datasets')
inventory = pd.read_csv(report_dir / 'visualnews_archive_inventory.csv')
samples = pd.read_parquet('../data/processed/visualnews/visualnews_smoke_manifest.parquet')
pickle_inspection = json.loads((report_dir / 'visualnews_pickle_static_inspection.json').read_text())
print(f'Archives audited: {len(inventory)}')
print(f'Representative article records: {len(samples)}')
print(f'Records with image references: {samples.image_url.notna().sum()}')
print(f'Verified local images: {(samples.download_status == "verified").sum()}')
display(inventory)
display(samples[['source_publication', 'article_id', 'caption', 'image_url', 'label', 'validation_status']].head(10))
pd.DataFrame(pickle_inspection)[['pickle', 'bytes', 'opcodes_scanned', 'unsafe_deserialization_opcodes_seen', 'deserialized']]
```
