# Multimodal integrity and leakage
Review the generated evidence before training or changing the sampling policy.

```python
import json
from pathlib import Path
import pandas as pd
root = Path('../data/research_dataset')
m = pd.read_parquet(root / 'manifests/unified_manifest.parquet')
leakage = json.loads((root / 'reports/leakage_report.json').read_text())
storage = json.loads((root / 'reports/storage_usage.json').read_text())
display(pd.DataFrame([leakage]))
display(pd.DataFrame([storage]))
f = m[m.source_dataset.eq('Fakeddit')]
assert f.groupby('image_hash').final_split.nunique().max() == 1
assert m.loc[m.source_dataset.eq('NewsCLIPpings'), 'fake_news_label'].isna().all()
print('Integrity checks passed.')
```
