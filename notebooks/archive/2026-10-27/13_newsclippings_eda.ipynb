# NewsCLIPpings EDA

EDA over official annotations. `falsified` indicates generated image-caption mismatch, not misinformation truthfulness.

```python
import pandas as pd

frame = pd.read_parquet('../data/processed/newsclippings/newsclippings_annotations_provisional.parquet')
print(f'Annotations: {len(frame):,}')
print(f'Resolved images: {frame.image_path.notna().sum():,}')
display(frame.groupby(['source_dataset', 'original_split', 'falsified']).size().rename('records').reset_index())
display(frame.isna().sum().rename('missing_values').to_frame())
```
