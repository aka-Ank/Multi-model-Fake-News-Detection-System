## Manual Convolution

For input `X` and filter `W`, the implemented convolution computes:

`Y[n, f, i, j] = sum_c,kh,kw X[n,c,i+kh,j+kw] * W[f,c,kh,kw] + b[f]`

The implementation uses an `im2col` transform: local image patches are rearranged into rows, then multiplied by flattened filters. Backpropagation maps column gradients back to image layout with `col2im`.
