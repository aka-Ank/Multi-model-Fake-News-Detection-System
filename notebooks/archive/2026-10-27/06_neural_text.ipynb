# 06 Neural Text Baseline

This notebook loads saved outputs for the manually implemented trainable embedding text model. It does not retrain the model.

## Model

The model uses a custom tokenizer and a training-only vocabulary with PAD and UNK IDs. Tokens are mapped into randomly initialized trainable embeddings, pooled with masked mean pooling over non-PAD tokens, then classified with Dense -> ReLU -> Dropout -> Dense -> Sigmoid.

```python
import json
from pathlib import Path

import pandas as pd

base = Path('../reports/experiments/neural_text')
multi = Path('../reports/experiments/neural_text_multiseed')
metrics = json.loads((base / 'metrics.json').read_text())
history = pd.read_csv(base / 'history.csv')
comparison = pd.read_csv(base / 'comparison.csv')
aggregate = pd.read_csv(multi / 'aggregate_metrics.csv')
metrics['validation_metrics']
```

## Embedding And Pooling Math

Embedding lookup maps token IDs to rows of `E ∈ R^(V × D)`. Backward accumulation uses repeated-index addition so repeated tokens accumulate gradients correctly.

Masked mean pooling computes `sum(valid embeddings) / number_valid_tokens`, ignoring PAD tokens so padding does not shrink or distort the representation.
