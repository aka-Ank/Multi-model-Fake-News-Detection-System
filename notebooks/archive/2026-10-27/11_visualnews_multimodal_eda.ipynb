# VisualNews Multimodal EDA

This is an evidence-limited EDA over the verified smoke manifest. It does not treat VisualNews articles as fake-news labels or assume that remote image URLs are downloaded.

```python
import pandas as pd

frame = pd.read_parquet('../data/processed/visualnews/visualnews_smoke_manifest.parquet')
frame['text_chars'] = frame.text.fillna('').str.len()
summary = pd.DataFrame([{
    'article_records': len(frame),
    'image_references': int(frame.image_url.notna().sum()),
    'verified_local_images': int(frame.image_path.notna().sum()),
    'labeled_records': int(frame.label.notna().sum()),
    'duplicate_content_hashes': int(frame.content_hash.duplicated().sum()),
    'median_text_characters': float(frame.text_chars.median()),
}])
display(summary)
display(frame.groupby('source_publication').agg(records=('sample_id', 'count'), image_references=('image_url', lambda x: x.notna().sum()), median_text_chars=('text_chars', 'median')))
frame[['text_chars']].describe()
```
